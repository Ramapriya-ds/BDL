```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
product_data = [
    "Laptop", "Mobile", "Headphones", "Keyboard", "Mouse",
    "Monitor", "Printer", "Office Chair", "Table", "Desk"
]

# 2. Category Data
category_data = [
    "Electronics", "Electronics", "Electronics", "Electronics",
    "Electronics", "Electronics", "Electronics",
    "Furniture", "Furniture", "Furniture"
]

# 3. Sales Data
sales_data = [
    55000, 44000, 25000, 18000, 12000,
    42000, 30000, 45000, 60000, 51000
]

# 4. Quantity Data
quantity_data = [
    5, 10, 20, 15, 25,
    7, 6, 8, 5, 6
]

# 5. Price Data
price_data = [
    55000, 22000, 2500, 1200, 500,
    14000, 15000, 7500, 12000, 8500
]

# 6. Date Data
date_data = [
    "2026-01-05", "2026-02-05", "2026-03-05",
    "2026-04-05", "2026-05-05", "2026-06-05",
    "2026-07-05", "2026-08-05", "2026-09-05",
    "2026-10-05"
]

# 7. Region Data
region_data = [
    "Chennai", "Madurai", "Coimbatore", "Chennai",
    "Trichy", "Madurai", "Coimbatore", "Chennai",
    "Trichy", "Coimbatore"
]

# 8. Profit Data
profit_data = [
    8250, 6600, 5000, 3600, 3000,
    6300, 4500, 9000, 12000, 10200
]

# 9. Discount Data
discount_data = [
    5, 10, 5, 8, 5,
    7, 6, 10, 5, 8
]

# 10. Customer Type Data
customer_data = [
    "Regular", "New", "Regular", "New", "Regular",
    "Regular", "New", "Regular", "New", "Regular"
]
```

```python
df = pd.DataFrame({

    "Product": product_data,
    "Category": category_data,
    "Sales": sales_data,
    "Quantity": quantity_data,
    "Price": price_data,
    "Date": date_data,
    "Region": region_data,
    "Profit": profit_data,
    "Discount": discount_data,
    "Customer_Type": customer_data

})
```

```python
print("\n================================================")
print("        RETAIL SALES DATASET")
print("================================================")

print(df)
```

```python
print("\n================================================")
print("        RETAIL SALES DATASET")
print("================================================")

print(df)
```

```python
print("\n================================================")
print("                DATA TYPES")
print("================================================")

print(df.dtypes)
```

```python
df["Date"] = pd.to_datetime(df["Date"])

# Check missing values

print("\n================================================")
print("             MISSING VALUES")
print("================================================")

print(df.isnull().sum())


# Check duplicate values

print("\n================================================")
print("             DUPLICATE VALUES")
print("================================================")

print("Duplicate Rows:", df.duplicated().sum())


# Remove duplicates if available

df = df.drop_duplicates()
```

```python
print("\n================================================")
print("             BASIC ANALYSIS")
print("================================================")

total_sales = df["Sales"].sum()
total_profit = df["Profit"].sum()
total_quantity = df["Quantity"].sum()
average_sales = df["Sales"].mean()
average_profit = df["Profit"].mean()

print("Total Sales       :", total_sales)
print("Total Profit      :", total_profit)
print("Total Quantity    :", total_quantity)
print("Average Sales     :", round(average_sales, 2))
print("Average Profit    :", round(average_profit, 2))
```

```python
product_sales = df.groupby("Product")["Sales"].sum()

print("\n================================================")
print("             PRODUCT-WISE SALES")
print("================================================")

print(product_sales)
```

```python


plt.bar(
    product_sales.index,
    product_sales.values,
    color="skyblue"
)

plt.title("Product-wise Sales Performance")
plt.xlabel("Product")
plt.ylabel("Sales")

plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python

plt.figure(figsize=(10, 6))

plt.plot(
    df["Date"],
    df["Sales"],
    marker="o",
    color="blue",
    linewidth=2
)

plt.title("Sales Trend Over Time")
plt.xlabel("Date")
plt.ylabel("Sales")

plt.grid(True)
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
category_sales = df.groupby("Category")["Sales"].sum()

print(category_sales)
```

```python
plt.figure(figsize=(7, 5))

plt.bar(
    category_sales.index,
    category_sales.values,
    color=["lavender", "skyblue"]
)

plt.title("Category-wise Sales")
plt.xlabel("Category")
plt.ylabel("Total Sales")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 6))

sns.scatterplot(
    data=df,
    x="Sales",
    y="Profit",
    hue="Category",
    s=120
)

plt.title("Sales vs Profit")
plt.xlabel("Sales")
plt.ylabel("Profit")

plt.show()
```

```python


plt.figure(figsize=(8, 5))

plt.hist(
    df["Sales"],
    bins=5,
    color="skyblue",
    edgecolor="black"
)

plt.title("Sales Distribution Histogram")
plt.xlabel("Sales Amount")
plt.ylabel("Frequency")

plt.grid(axis="y", alpha=0.3)

plt.tight_layout()
plt.show()
```

```python


# Category-wise total sales
category_sales = df.groupby("Category")["Sales"].sum()

# Pie Chart
plt.figure(figsize=(7, 7))

plt.pie(
    category_sales.values,
    labels=category_sales.index,
    autopct="%1.1f%%",
    startangle=90,
    colors=["purple", "black"]
)

plt.title("Sales Distribution by Category")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.boxplot(
    y=df["Sales"],
    color="skyblue"
)

plt.title("Sales Distribution - Box Plot")
plt.ylabel("Sales")

plt.show()
```
